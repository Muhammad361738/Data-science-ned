# Data Visualization Methods for Car Dataset


Below are some common plotting methods for visualizing the car dataset.

1. **Pie Chart for Percentage of Customers by `Churn`**:
Shows the distribution of customers by `Churn`.
```python
gb = df.groupby('Churn').agg({'Churn': 'count'})

plt.pie(gb['Churn'], labels=gb.index, autopct='%1.2f%%')
plt.title('Percentage of customer by Churn')
plt.show()
```

2. **Bar Chart for Churn by Gender**:
Visualizes the `Churn` status by gender with count bars.
```python
plt.figure(figsize=(5, 5))
ax = sns.countplot(x='gender', data=df, hue='Churn')

ax.bar_label(ax.containers[0])
plt.title('Churn by Gender')
plt.show()
```

3. **Box Plot for Price by `carbody`**:
Shows the distribution of `price` based on the `carbody` type.
```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='carbody', y='price', data=df)
plt.title('Price Distribution by Car Body Type')
plt.show()
```

4. **Histogram for `horsepower` Distribution**:
Visualizes the distribution of horsepower.
```python
plt.figure(figsize=(10, 6))
sns.histplot(df['horsepower'], bins=20, kde=True)
plt.title('Horsepower Distribution')
plt.show()
```

5. **Scatter Plot for `horsepower` vs `price`**:
Shows the relationship between horsepower and price.
```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='horsepower', y='price', data=df)
plt.title('Horsepower vs Price')
plt.show()
```

6. **Pairplot for Multiple Variables**:
Plots pairwise relationships between selected numerical features.
```python
sns.pairplot(df[['price', 'horsepower', 'enginesize', 'curbweight']])
plt.title('Pairplot of Price, Horsepower, Engine Size, and Curb Weight')
plt.show()
```

7. **Correlation Heatmap for Numerical Columns**:
Displays the correlation between numerical features.
```python
plt.figure(figsize=(10, 8))
sns.heatmap(df[['price', 'horsepower', 'enginesize', 'curbweight']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

8. **Bar Chart for Number of Cars by `fueltype`**:
Visualizes the count of cars by fuel type.
```python
plt.figure(figsize=(8, 6))
sns.countplot(x='fueltype', data=df)
plt.title('Number of Cars by Fuel Type')
plt.show()
```

9. **Count Plot for `carbody`**:
Shows the count of cars by `carbody` type.
```python
plt.figure(figsize=(8, 6))
sns.countplot(x='carbody', data=df)
plt.title('Number of Cars by Car Body Type')
plt.show()
```

10. **Scatter Plot for `curbweight` vs `enginesize`**:
Shows the relationship between curb weight and engine size.
```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='curbweight', y='enginesize', data=df)
plt.title('Curb Weight vs Engine Size')
plt.show()
```
