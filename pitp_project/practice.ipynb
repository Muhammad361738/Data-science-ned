# 1. **Highest Price**: Finds the highest price in the dataset.


```python
highest_price = df['price'].max()
print("Heighest price = ",highest_price)
```

# 2. **Lowest Price**: Finds the lowest price in the dataset.


```python
lowest_price = df['price'].min()
print("Lowest price = ",lowest_price)
```

# 3. **Average Price**: Calculates the average price of all cars.


```python
average_price = df['price'].mean()
print("Average price of All cars = ",average_price)
```

# 4. **Car with Highest Horsepower**: Finds the car with the highest horsepower.


```python
car_highest_hp = df.loc[df['horsepower'].idxmax()]
print("Heighest Horse Power = ",car_highest_hp)
```

## Transforming Data



The price column should ideally have a numerical data type (e.g., float or int), but if it has an object datatype, it could be due to non-numeric values like empty strings or text in the CSV file.

To fix this, we first need to replace empty strings with 0 or NaN, depending on the context. For example, if price is missing and we know related columns (like enginesize or horsepower) indicate a non-usable car, we can set missing prices to 0.

After cleaning, we can convert the price column to float using:



```python
df['price'] = df['price'].replace('', 0)  # Or use np.nan if preferred
df['price'] = df['price'].astype(float)
```

# Car body analysis 

```python
ax = sns.countplot(x='carbody', data=df)
ax.bar_label(ax.containers[0])
plt.title('Count of Cars by Body Type')
plt.show()
# What This Will Show:
# A bar chart showing the number of cars for each type of carbody (e.g., sedan, hatchback, wagon, etc.), with count labels on top of each bar.

# Let me know if you'd rather visualize another column like fueltype or drivewheel
```

```python
gb = df.groupby('fueltype').agg({'fueltype': 'count'})

plt.pie(gb['fueltype'], labels=gb.index, autopct='%1.2f%%')
plt.title('Percentage of Cars by Fuel Type')
plt.show()
# A pie chart showing the proportion of cars that use gas vs diesel.

# Let me know if you want to use another column like carbody, aspiration, or drivewheel — I can tweak the code accordingly!

```

## Drive wheel  based on the fuel type 

```python
plt.figure(figsize=(5, 5))
ax = sns.countplot(x='drivewheel', data=df, hue='fueltype')

ax.bar_label(ax.containers[0])
plt.title('Drive Wheel Type by Fuel Type')
plt.show()
```
