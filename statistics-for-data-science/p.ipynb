🔹 Q1: Z-Test (One-Sample)
A factory claims that their light bulbs last on average 1200 hours. A sample of 40 bulbs has a mean of 1170 hours with a standard deviation of 80 hours.

At 5% significance level, test the claim using Z-test.



```python
import pandas as pd
import numpy as np
from scipy.stats import norm
mu=1200
n=40
smple=1170
sigma=80
alpha=0.05

z=(smple-mu)/(sigma/np.sqrt(n))
print("Z value ",round(z,3))

z_critical=norm.ppf(1-alpha/2)
print("Z_critical value ",z_critical)
p_value = 2 * norm.cdf(z)  # since z is negative
print("P-value:", p_value) 

if z<z_critical:
    print("Reject null hypthesis")
else:
    print("fail to reject null hypothesis")
    
```

 Q2: T-Test (One-Sample)
The average cholesterol level in a city is believed to be 190 mg/dL. A sample of 10 patients has levels:


198,202,187,210,190,185,199,200,188,192

Use t-test at α = 0.05.



```python
from scipy.stats import ttest_1samp

sample=[198,202,187,210,190,185,199,200,188,192]
t_tes,p_value=ttest_1samp(sample,190)

print("T Test Value",round(t_tes,2))
print("p_value",round(p_value,2))

if p_value < 0.05:
        print("Rject to null values ")
else:
    print("Fail to rejet null values")
```

 Q3: Chi-Square Test of Independence
Gender	A	B	C
Male	20	15	25
Female	10	30	20

Check if gender and product preference are independent

```python
import pandas as pd
from scipy.stats import chi2_contingency
import numpy as np
observed=np.array([
    [20,15,25],
    [10,30,20]
])
df=pd.DataFrame(observed,index=["Male","Female"],columns=["A","B","C"])

chi2,p_value,dof,expected=chi2_contingency(df)
print("Chaii square",chi2)
print("P value is",p_value)
print("Dof",dof)
print("Expected",expected)
if p_value < 0.05:
    print("Reject null hypothesis")
else:
    print("Fail to reject null hypothesis")
```

```python
import pandas as pd
from sklearn.feature_selection import VarianceThreshold
import numpy as np


df = pd.DataFrame({
    'A': [5.1, 5.2, 5.0, 5.1, 5.2, 5.0],
    'B': [100, 102, 98, 99, 100, 101],
    'C': [10, 10, 10, 10, 10, 10]  
})


selector = VarianceThreshold(threshold=0.01)
df_var = selector.fit_transform(df)


retained_columns = df.columns[selector.get_support()]
df_cleaned = pd.DataFrame(df_var, columns=retained_columns)

if 'B' in df_cleaned.columns:
    Q1 = df_cleaned['B'].quantile(0.25)
    Q3 = df_cleaned['B'].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_cleaned = df_cleaned[(df_cleaned['B'] >= lower_bound) & (df_cleaned['B'] <= upper_bound)]

#
print("Cleaned DataFrame:\n", df_cleaned)
```
