#### ✅ Task 1: Student Performance Analysis
Step 1 – Pandas:
Create a DataFrame of 10 students with the following columns:

Name

Maths Marks

Science Marks

English Marks

You can manually create it or use NumPy's random integers (between 50 and 100) to auto-generate marks.

Step 2 – NumPy:

Convert only the marks columns into a NumPy array.

Calculate average marks for each student (row-wise).

Find the student with the highest total marks.




```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({"Name":["Talha","Zain","Waqar","Salman","Zeeshan"],
                   "Math":np.random.randint(50,101,5),
                   "Science":np.random.randint(50,101,5),
                   "English":np.random.randint(50,101,5),
                   
                   })
df
```

```python
df["total_marks"]=np.sum(df[["English","Science","Math"]], axis=1)
df["Avgerage"]=np.mean(df[["English","Science","Math"]], axis=1)
top_std = df["Avgerage"].max()
print(df)
print()
print(top_std)
```

## Step 3 – Matplotlib:

#### Plot a bar chart comparing total marks of all students.

#### Add student names as x-axis labels.


```python
plt.bar(df["Name"], df["total_marks"], color="orange", label = "Student marks compersion")
plt.title("Student Final Exam Result Compersion")
plt.xlabel("Student Name")
plt.ylabel("Total Marks")
plt.grid()
plt.legend()
plt.show()
```

## ✅ Task 2: Salary Data Analysis (CSV Based)
Step 1 – Pandas:
Create or load a CSV file with the following fields for 15 employees:

Name

Department (IT, HR, Marketing, Finance)

Age

Salary



```python
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Zara", "Usman", "Areeba", "Bilal", "Hira", "Danish", "Mehwish", "Omar", "Sana", "Fahad", "Noor", "Hassan"],
    "Department": ["IT", "HR", "Marketing", "Finance", "IT", "HR", "Marketing", "Finance", "IT", "HR", "Marketing", "Finance", "IT", "HR", "Marketing"],
    "Age": [28, 32, 26, 30, 35, 29, 31, 27, 24, 33, 28, 26, 36, 30, 25],
    "Salary": [75000, 65000, 55000, 72000, 80000, 60000, 58000, 70000, 68000, 67000, 54000, 71000, 82000, 66000, 56000]
}

df = pd.DataFrame(data)
df
```

### Step 2 – Pandas + NumPy:

Calculate average salary department-wise using groupby.

Use NumPy to calculate standard deviation of all salaries.


```python
avg=df.groupby("Department")["Salary"].mean()
pd.DataFrame(avg)
```

```python
std_all_sal = df["Salary"].std()
std_all_sal
```


### Step 3 – Matplotlib:

Create a pie chart of how many employees are in each department.

Create a histogram of the salaries.



```python
plt.pie(df["Department"].value_counts(), labels=df["Department"].value_counts().index, colors=["purple","gold","green","yellow"],autopct="%1.1f%%")
plt.title("Employee ratio in each deparment ")
plt.figure(figsize=(8, 6))
plt.show()
```

```python
plt.hist(df["Salary"] , bins=20 , color = "green", edgecolor="black")
plt.xlabel("Salary")
plt.ylabel("Number of employess")
plt.title("Histograam of the salaries ")
plt.show()
```

# ✅ Task 3: Temperature Trends

Step 1 – NumPy:
Create temperature data for 7 days using:

np.random.randint(20, 40, size=7)




```python
temperature = np.random.randint(20, 40, size=7)
```

### Step 2 – Pandas:
Create a DataFrame with:

Day (Mon to Sun)

Temperature

Feels Like (Temperature + some random offset)


```python

# Days of the week
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

# Random offset for "Feels Like" temperature (between -2 to +2)
offset = np.random.randint(-2, 3, size=7)
feels_like = temperature + offset

# Create DataFrame
df = pd.DataFrame({
    'Day': days,
    'Temperature': temperature,
    'Feels Like': feels_like
})

print(df)
```

### Step 3 – Matplotlib:

Create a line chart of Temperature vs Day

Add another line for “Feels Like” temperature

Add legend, title, and grid


```python
plt.plot(df["Day"],df["Temperature"],df["Feels Like"], marker="o",label="Temperatue graph")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Temperature according to week days ")
plt.grid()
plt.show()
```

```python
plt.plot(df["Day"], df["Temperature"], marker="o", label="Temperature")
plt.plot(df["Day"], df["Feels Like"], marker="x", label="Feels Like")
plt.legend()
```
