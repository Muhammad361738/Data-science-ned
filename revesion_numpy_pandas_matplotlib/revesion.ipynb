Create a DataFrame of students with columns: Name, Age, Marks, and City.

Filter students whose marks are greater than 80 and city is "Karachi".

Create a DataFrame of employees with: Name, Department, Salary.

✅ Question 3:
Create a DataFrame with some missing values (NaN) in Age or Marks.

Use functions to fill missing values with average.

Then drop rows where all values are missing.



```python
import pandas as pd 
```

```python
df = pd.DataFrame({"Name" : ["Talha","Zain","Waqar","Salman","Sufyan"],
                   "Age":[20,21,22,20,24],
                   "Maarks":[96,90,85,98,97],
                   "City":["Karachi","peshawer","Swat","Karachi","Karachi"]})
df
```

```python
df1=df[(df["City"]=="Karachi")&(df["Maarks"] >= 80)]
df1
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Department': ['HR', 'IT', 'Finance', 'Marketing'],
    'Salary': [50000, 60000, 55000, 52000]
}

# Create the DataFrame
df = pd.DataFrame(data)
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
data = {
    'Name': ['Ali', 'Sara', 'John', 'Zara'],
    'Age': [25, np.nan, 22, 24],
    'Marks': [85, 90, np.nan, 88]
}

df = pd.DataFrame(data)


df["Age"].fillna(df["Age"].mean(),inplace=True)
df["Marks"].fillna(df["Marks"].mean(),inplace=True)
df
```

## 🔷 Matplotlib – Practice Questions with Detail


```python
import matplotlib.pyplot as plt 
```

✅ Question 1:
Create a line plot showing temperature over 7 days.

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temperature = [30, 32, 33, 31, 29, 35, 36]

plt.plot(days,temperature)
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Temperature graph of the geek ")
plt.grid(True)
plt.show()
```

✅ Question 2:
Create a bar chart showing the sales of 4 products: A, B, C, D.
## Add bar color and display value labels on top of bars.



```python
products = ['A', 'B', 'C', 'D']
sales = [120, 90, 150, 100]

plt.bar(products , sales, color = "purple", label = "product sales ")
plt.title("Sales per product ")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.legend()
plt.grid(color = "gray", linestyle = ":", linewidth = 1)
plt.show()
```

✅ Question 3:
Create a pie chart showing percentage of students in different courses:

Python: 40

JavaScript: 25

Java: 20

C++: 15



```python
sub = ["Python","JavaScript","Java","C++"]
std = [40,25,20,15]
plt.pie(std , labels=sub , colors=["purple","orange","gold","green"] , autopct="%1.1f%%")
plt.title("Percentage of student enroll each subject")
plt.show()
```
