Create a DataFrame of students with columns: Name, Age, Marks, and City.

Add at least 5 rows.

Filter students whose marks are greater than 80 and city is "Karachi".

Create a DataFrame of employees with: Name, Department, Salary.

Create a DataFrame with some missing values (NaN) in Age or Marks.

Use functions to fill missing values with average.

Then drop rows where all values are missing.



```python
import pandas as pd 
```

```python
df = pd.DataFrame({"Name" : ["Talha","Zain","Waqar","Salman","Sufyan"],
                   "Age":[20,21,22,20,24],
                   "Maarks":[96,90,85,98,97],
                   "City":["Karachi","peshawer","Swat","Karachi","Karachi"]})
df
```

```python
df1=df[(df["City"]=="Karachi")&(df["Maarks"] >= 80)]
df1
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Department': ['HR', 'IT', 'Finance', 'Marketing'],
    'Salary': [50000, 60000, 55000, 52000]
}

# Create the DataFrame
df = pd.DataFrame(data)
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
data = {
    'Name': ['Ali', 'Sara', 'John', 'Zara'],
    'Age': [25, np.nan, 22, 24],
    'Marks': [85, 90, np.nan, 88]
}

df = pd.DataFrame(data)


df["Age"].fillna(df["Age"].mean(),inplace=True)
df["Marks"].fillna(df["Marks"].mean(),inplace=True)
df
```
