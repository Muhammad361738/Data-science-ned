### ✅ Task 5: Statistical Operations
Create a 5x5 array with values from 1 to 25.
Now calculate:

Mean

Median

Standard Deviation

Variance



```python
ar = np.random.randint(1,101,size=(5,5))

print(ar)

for i in ar:
    print(i.mean())
    # print(i.median())
    print(i.std())
    print(i.var(),"\n")
```
