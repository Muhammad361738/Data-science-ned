📊 Netflix Content Analysis - Project Guide

🎯 Project Overview:
Analyze Netflix's content library to uncover insights about genre distribution, content ratings, release trends, and more.

📁 Dataset Details:
- Source: Netflix Movies and TV Shows Dataset on Kaggle
- Format: CSV
- Fields:
  • show_id
  • type (Movie/TV Show)
  • title
  • director
  • cast
  • country
  • date_added
  • release_year
  • rating
  • duration
  • listed_in (Genres)
  • description

🧰 Tools Required:
- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn (optional)

🗂️ Step-by-Step Instructions:

1. Data Loading and Initial Exploration
   - Import libraries
   - Load dataset with Pandas
   - Display first few rows
   - Check for null values and data types

2. Data Cleaning
   - Handle missing values
   - Convert 'date_added' to datetime
   - Extract year and month from 'date_added'

3. Exploratory Data Analysis (EDA)
   - Content type distribution
   - Top genres
   - Releases per year
   - Content ratings
   - Duration distribution

4. Visualization
   - Bar charts (types, ratings)
   - Line graph (content addition over years)
   - Heatmap (correlation)
   - Pie charts (genre distribution)

5. Advanced Analysis (Optional)
   - Ratings vs. genres
   - Top directors/actors
   - Country-wise content analysis

📌 Deliverables:
- Jupyter Notebook (`netflix_analysis.ipynb`) containing:
  - Data loading and cleaning
  - EDA with visualizations
  - Insights and observations
- Visualizations saved as images (optional)
